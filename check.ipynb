```python
import os
import pandas as pd
import re
import json
from collections import defaultdict

# 📂 Folder containing cleaned student files
input_folder = r"cleaned_outputs_final"

# 🧠 Helper functions
def extract_name(filename):
    return filename.replace('.csv', '').strip().lower()

def normalize_name(name):
    name = str(name).strip().lower()
    name = re.sub(r'\s+', ' ', name)
    return name

# 📁 Lists to track
files = [f for f in os.listdir(input_folder) if f.endswith('.csv')]
have_file = set(extract_name(f) for f in files)
adjacency_list = defaultdict(set)
not_have_file = set()
name_to_file = {}

# 🔁 Process each file to build graph
for file in files:
    file_path = os.path.join(input_folder, file)
    try:
        df = pd.read_csv(file_path)

        owner = extract_name(file)
        name_to_file[owner] = file

        if 'full_name' in df.columns:
            adjacent_people = df['full_name'].dropna().map(normalize_name)
        elif 'first name' in df.columns and 'last name' in df.columns:
            adjacent_people = (df['first name'] + ' ' + df['last name']).map(normalize_name)
        else:
            continue

        for person in adjacent_people:
            if person == owner:
                continue
            adjacency_list[owner].add(person)
            if person not in have_file:
                not_have_file.add(person)

    except Exception as e:
        print(f"❌ Error reading {file}: {e}")

# 🔁 Build reverse connections for those without file
for unknown in not_have_file:
    for known in list(adjacency_list.keys()):
        if unknown in adjacency_list[known]:
            adjacency_list[unknown].add(known)

# 🧾 Save adjacency list to JSON
json_path = os.path.join(input_folder, 'adjacency_list.json')
with open(json_path, 'w', encoding='utf-8') as f:
    json.dump({k: list(v) for k, v in adjacency_list.items()}, f, indent=4)
print(f"✅ Adjacency list saved to JSON → {json_path}")

# 🧾 Save adjacency list to TXT with ':' instead of '→'
txt_path = os.path.join(input_folder, 'adjacency_list.txt')
with open(txt_path, 'w', encoding='utf-8') as f:
    for k, v in adjacency_list.items():
        f.write(f"{k}: {', '.join(v)}\n")
print(f"✅ Adjacency list saved to TXT → {txt_path}")

# 📊 Final stats
print(f"\n👥 Total students with file: {len(have_file)}")
print(f"👀 Students without file (extracted from others): {len(not_have_file)}")
```

```python
import os
import pandas as pd

# 📂 Input folders
original_folder = r"new_renamed_files"  # path to original files
cleaned_folder = r"check_cleaned_outputs_final"  # path to cleaned files

# 📄 Get only file names (no extensions) from original
original_files = set()
for file in os.listdir(original_folder):
    if file.endswith(('.csv', '.xlsx')):
        name = os.path.splitext(file)[0]
        original_files.add(name.lower())

# 📄 Get only file names (no extensions) from cleaned folder
cleaned_files = set()
for file in os.listdir(cleaned_folder):
    if file.endswith('.csv'):
        name = os.path.splitext(file)[0]
        cleaned_files.add(name.lower())

# ❌ Find missing files (in original but not in cleaned)
missing_files = original_files - cleaned_files

# 📝 Save or print the list
print(f"\n📛 Total missing cleaned files: {len(missing_files)}")
for f in sorted(missing_files):
    print(f"❌ {f}")

# Optional: Save to CSV
pd.DataFrame(sorted(missing_files), columns=["Missing_File_Name"]).to_csv("missing_cleaned_files.csv", index=False)
print("\n✅ Missing file list saved as: missing_cleaned_files.csv")
```

```python
import os
import shutil

# 💾 Paths
original_folder = r"new_renamed_files"
cleaned_folder = r"check_cleaned_outputs2_final"
missing_output_folder = r"missing_cleaned_files2"

# 📁 Make sure missing folder exists
os.makedirs(missing_output_folder, exist_ok=True)

# 📋 Get all original filenames (only .csv and .xlsx)
original_files = [
    f for f in os.listdir(original_folder)
    if os.path.isfile(os.path.join(original_folder, f)) and f.lower().endswith(('.csv', '.xlsx'))
]

# 📋 Get all cleaned filenames
cleaned_files = [
    f for f in os.listdir(cleaned_folder)
    if os.path.isfile(os.path.join(cleaned_folder, f)) and f.lower().endswith('.csv')
]

# 🧠 Normalize cleaned names (without extension)
cleaned_basenames = set(os.path.splitext(f)[0].lower() for f in cleaned_files)

missing = []

for file in original_files:
    base = os.path.splitext(file)[0].lower()
    if not any(base in cleaned for cleaned in cleaned_basenames):
        # 🚚 Copy missing file to new folder
        src = os.path.join(original_folder, file)
        dst = os.path.join(missing_output_folder, file)
        shutil.copy2(src, dst)
        missing.append(file)

# ✅ Summary
print(f"📦 Total original files: {len(original_files)}")
print(f"✅ Cleaned files found: {len(cleaned_files)}")
print(f"❌ Missing cleaned files: {len(missing)}")
print(f"📂 Missing files saved to: {missing_output_folder}")
```
